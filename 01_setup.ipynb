# 01: Setup & Dataset Download
NasalSeg 資料集下載（Zenodo）。只需要跑一次。

```python
# ============================================================
# 下載 NasalSeg Dataset from Zenodo (原版方式)
# ============================================================
import os, shutil, zipfile, pathlib, sys, subprocess

print('🚀 Downloading NasalSeg Dataset from Zenodo...')

work = pathlib.Path('/workspace/zenodo_tmp')
work.mkdir(parents=True, exist_ok=True)

# 方法 1: zenodo-get
print('📦 Installing zenodo-get...')
subprocess.run([sys.executable,'-m','pip','install','zenodo-get','-q'], check=False)
print('⬇️  Downloading with zenodo-get...')
result = subprocess.run(
    [sys.executable,'-m','zenodo_get','13893419','-o',str(work)],
    capture_output=True, text=True)

if result.returncode == 0:
    print('✅ zenodo-get download successful')
else:
    print('⚠️  zenodo-get failed, trying wget...')
    main_file = work / 'NasalSeg.zip'
    url = 'https://zenodo.org/records/13893419/files/NasalSeg.zip?download=1'
    if main_file.exists() and main_file.stat().st_size > 200_000_000:
        print(f'✅ File already exists: {main_file}')
    else:
        subprocess.run(['wget','--progress=bar','--show-progress','-O',str(main_file),url])

# 解壓縮
zip_files = list(work.glob('*.zip'))
if not zip_files:
    print('❌ No zip files found')
    print('Manually download: https://zenodo.org/records/13893419/files/NasalSeg.zip')
    print(f'Place it in: {work}')
else:
    for zf in zip_files:
        print(f'📦 Extracting {zf.name} ({zf.stat().st_size/1e6:.1f} MB)...')
        with zipfile.ZipFile(zf,'r') as z:
            print(f'   {len(z.namelist())} files')
            z.extractall(work)
        print('   ✅ Done')

# 確認結構
images_dir = work / 'NasalSeg' / 'images'
labels_dir = work / 'NasalSeg' / 'labels'
if images_dir.exists() and labels_dir.exists():
    img_count = len(list(images_dir.glob('*.nrrd')))
    seg_count = len(list(labels_dir.glob('*.nrrd')))
    print(f'✅ Images: {img_count}, Labels: {seg_count}')
    if img_count == 130:
        print('✅ All 130 cases found!')
    else:
        print(f'⚠️  Expected 130, found {img_count}')
else:
    print('⚠️  Expected directory structure not found')
    print(f'   Looked for: {images_dir}')
```
